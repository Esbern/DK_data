# 03 · Build an interactive Danish map

This notebook combines public Danish boundaries with Folium for web-style interactive visualization.

```python
# If needed, install dependencies:
# %pip install geopandas folium

import geopandas as gpd
import folium
```

```python
url = "https://github.com/wmgeolab/geoBoundaries/raw/main/releaseData/gbOpen/DNK/ADM1/geoBoundaries-DNK-ADM1.geojson"
regions = gpd.read_file(url).to_crs(epsg=4326)

# Center roughly on Denmark
m = folium.Map(location=[56.2, 10.0], zoom_start=6, tiles="cartodbpositron")
```

```python
preferred_name_fields = ['shapeName', 'name', 'NAME_1', 'region', 'Region']
name_field = next((col for col in preferred_name_fields if col in regions.columns), None)

geojson_kwargs = {'name': 'Danish regions'}
if name_field:
    geojson_kwargs['tooltip'] = folium.GeoJsonTooltip(fields=[name_field], aliases=['Region'])

folium.GeoJson(regions, **geojson_kwargs).add_to(m)

folium.LayerControl().add_to(m)
m
```
