# 02 · Access OpenStreetMap data for Denmark

This notebook shows how to pull Danish street and place data using OSMnx.

```python
# If needed, install dependencies:
# %pip install osmnx matplotlib

import osmnx as ox
import matplotlib.pyplot as plt
```

```python
place_name = "Copenhagen, Denmark"

# Download drivable street network
G = ox.graph_from_place(place_name, network_type="drive")

# Download parks as example polygon data
parks = ox.features_from_place(place_name, {"leisure": "park"})
parks.head()
```

```python
fig, ax = ox.plot_graph(G, node_size=0, edge_linewidth=0.5, show=False, close=False)
if not parks.empty:
    parks.plot(ax=ax, color='green', alpha=0.4)
ax.set_title("Copenhagen road network with parks")
plt.show()
```
