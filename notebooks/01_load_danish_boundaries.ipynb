# 01 · Load Danish administrative boundaries

This notebook shows how to access Danish spatial boundary data from a public GeoJSON endpoint and make a first map with GeoPandas.

```python
# If needed, install dependencies in your environment:
# %pip install geopandas matplotlib

import geopandas as gpd
import matplotlib.pyplot as plt
```

```python
# Danish ADM1 (regions) from geoBoundaries
url = "https://github.com/wmgeolab/geoBoundaries/raw/main/releaseData/gbOpen/DNK/ADM1/geoBoundaries-DNK-ADM1.geojson"
regions = gpd.read_file(url)
regions.head()
```

```python
ax = regions.plot(figsize=(8, 8), edgecolor="black", alpha=0.6)
ax.set_title("Danish regions (ADM1)")
ax.set_axis_off()
plt.show()
```
